```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2
    def sub(self):
        return self.num1 - self.num2
    def mul(self):
        return self.num1 * self.num2
    def div(self):
        return self.num1 / self.num2

print("===== CALCULATOR =====")
print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
choice=int(input("Enter Your Choice : "))
    
num1 = int(input("Enter First Number: "))
num2 = int(input("Enter Second Number: "))
calc = Calculator(num1, num2)

if choice==1:
    print(f"Addition of {num1} and {num2} is {calc.add()}")
elif choice==2:
    print(f"Subtraction of {num1} and {num2} is {calc.sub()}")
elif choice==3:
    print(f"Multiplication of {num1} and {num2} is {calc.mul()}")
elif choice==4:
    print(f"Division of {num1} and {num2} is {calc.div()}")
else:
    print("INVALID CHOICE")
```
