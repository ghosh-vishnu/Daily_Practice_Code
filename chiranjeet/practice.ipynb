```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2
    def sub(self):
        return self.num1 - self.num2
    def mul(self):
        return self.num1 * self.num2
    def div(self):
        return self.num1 / self.num2

print("===== CALCULATOR =====")
print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
choice=int(input("Enter Your Choice : "))
    
num1 = int(input("Enter First Number: "))
num2 = int(input("Enter Second Number: "))
calc = Calculator(num1, num2)

if choice==1:
    print(f"Addition of {num1} and {num2} is {calc.add()}")
elif choice==2:
    print(f"Subtraction of {num1} and {num2} is {calc.sub()}")
elif choice==3:
    print(f"Multiplication of {num1} and {num2} is {calc.mul()}")
elif choice==4:
    print(f"Division of {num1} and {num2} is {calc.div()}")
else:
    print("INVALID CHOICE")
```

```python
class BankAccount:
    def __init__(self,account_holder,account_number,balance):
        self.aaccount_holder=account_holder
        self.account_number=account_number
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print(f"Deposite Successfully of rs{amount}")

    def withdraw(self,amount):
        if self.balance>=amount:
            self.balance-=amount
            print(f"Withdrawl Successfully rs {amount}")
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print(f"Account Holder Name : {self.aaccount_holder}")
        print(f"Account Number : {self.account_number}")
        print(f"Available balance on your account rs {self.balance}\n")

b1=BankAccount("Rahul",1111111111,10000)
b2=BankAccount("Chiranjeet",2222222222,5000)
b3=BankAccount("Jitu",3333333333,6000)

print("======== Account 1 =========")
b1.deposit(5000)
b1.withdraw(2000)
b1.check_balance()

print("======== Account 2 =========")
b2.deposit(6000)
b2.withdraw(4000)
b2.check_balance()

print("======== Account 3 =========")
b3.deposit(5000)
b3.withdraw(200000)
b3.check_balance()
```

```python
class Car:

    company="Toyota"

    def __init__(self,model,color,price):
        self.model=model
        self.color=color
        self.price=price

    def start(self):
        print(f"{self.model} is Starting.")

    def stop(self):
        print(f"{self.model} is Stopped.\n")

    def display(self):
        print("======= CAR DETAILS =======")
        print(f"Company : {Car.company}")
        print(f"Model   : {self.model}")
        print(f"Color   : {self.color}")
        print(f"Price   : {self.price}rs")

c1=Car("Fortuner","Blace",5000000)
c2=Car("Innova","White",2500000)
c3=Car("Glanza","Red",1000000)

c1.display()
c1.start()
c1.stop()

c2.display()
c2.start()
c2.stop()

c3.display()
c3.start()
c3.stop()
```

```python
class Patient:
    def __init__(self,patient_name,age,disease,doctor_name):
        self.patient_name=patient_name
        self.age=age
        self.disease=disease
        self.doctor_name=doctor_name

    def display_patient(self):
        print(f"Name of the Patient : {self.patient_name}")
        print(f"Age of the Patient : {self.age}")
        print(f"Disease of the Patient : {self.disease}")
        print(f"Doctor Name : {self.doctor_name}\n")

    def admit(self,add):
        print(f"Patient Number {add} Admitted Successfully.\n")

    def discharge(self,add):
        print(f"Patient {add} Discharged.")

p1=Patient("Ram",36,"Typhoid","Rahul")
p2=Patient("Mohan",45,"Malaria","Chiranjeet")
p3=Patient("Sohan",53,"Cancer","Jitu")

print("========= Patient 1 ==========\n")
p1.admit(1)
p1.display_patient()

print("========= Patient 2 ==========\n")
p2.admit(2)
p2.display_patient()

print("========= Patient 3 ==========\n")
p3.admit(3)
p3.display_patient()
p3.discharge(3)
```
